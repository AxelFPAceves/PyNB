### Funciones con parámetros

#### Formato estándar:

```python
def <nombre>(<param_1>, <param_2>, ..., <param_n>):
  <instrucciones>
```

````python
def calcular_promedio(notas):
    """Calcula el promedio de una lista de notas.

    Args:
        notas: Una lista de números que representan las notas.

    Returns:
        El promedio de las notas, redondeado a una casilla decimal.
        Retorna 0 si la lista está vacía.
    """
    if not notas:
        return 0
    return int(round(sum(notas) / len(notas), 1))


# Ejemplo de uso

notas_estudiante = [9, 8.5, 6.5]
promedio = calcular_promedio(notas_estudiante)
print(f"El promedio del estudiante es: {promedio}")


def saludar(nombre):
    """Imprime un saludo personalizado.

    Args:
        nombre: El nombre de la persona a saludar.
    """
    print(f"Hola, {nombre}!")


# Ejemplo de uso
saludar("Juan")
````

## 2.3 Funciones que retornan valores

Retomando la actividad anterior, podemos retornar y guardar el valor del promedio de la siguiente forma:

````python
# Notas del estudiante
notas = [8.5, 2.0, 6.0, 1.0]
````

````python
def promedioEstudiante(*notas):
  if not notas:
    print("No se proporcionaron notas.")
    return 0

  promedio = sum(notas) / len(notas)

  if promedio > 6:
    print("Aprobado")
  else:
    print("Reprobado")

  return round(promedio, 2)

# Ejemplo de uso
notas_estudiante = [8.5, 9.0, 6.0, 10.0, 7.5]
promedio = promedioEstudiante(*notas_estudiante)
print(f"El promedio del estudiante es: {promedio}")
````

````python
def calcular_promedio_ponderado():
  notas = []
  pesos = [3, 2, 5]

  for i in range(3):
    while True:
      try:
        nota = float(input(f"Ingrese la nota {i+1}: "))
        if 0 <= nota <= 10:
          notas.append(nota)
          break
        else:
          print("La nota debe estar entre 0 y 10. Intente nuevamente.")
      except ValueError:
        print("Ingrese un valor numérico válido.")

  promedio_ponderado = sum(nota * peso for nota, peso in zip(notas, pesos)) / sum(pesos)
  return promedio_ponderado

promedio = calcular_promedio_ponderado()
print(f"El promedio ponderado es: {promedio}")
````

````python
dict_ciu = {
    sub[0] : len(sub)         # clave = nombre del estado (primer elemento)
                              # valor = # de apariciones (longitud de la sublista)
    for sub in lista_intermedia
}
print(dict_ciu)
````
